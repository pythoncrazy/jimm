```python
import jax

jax.config.update("jax_num_cpu_devices", 8)
```

```python
import jax.numpy as jnp
from flax import nnx
from jax.experimental import mesh_utils
from jax.sharding import Mesh

devices = mesh_utils.create_device_mesh((jax.device_count() // 2, 2))
mesh = Mesh(devices, ("fsdp"))
jax.set_mesh(mesh)
```

```python
sharding = jax.NamedSharding(mesh, jax.P("fsdp", "fsdp"))

A = jnp.zeros((1024, 1024), device=sharding)
jax.debug.visualize_array_sharding(A)
```

```python
from jimm.models.vit import VisionTransformer

HF_MODEL_NAME = "google/vit-large-patch16-384"
model = VisionTransformer.from_pretrained(HF_MODEL_NAME, use_pytorch=True, mesh=mesh, dtype=jnp.bfloat16)
nnx.display(model)
```

```python
from jimm.models.clip import CLIP

HF_MODEL_NAME = "openai/clip-vit-base-patch32"
USE_PYTORCH = True
model = CLIP.from_pretrained(HF_MODEL_NAME, use_pytorch=USE_PYTORCH)
nnx.display(model)
```
